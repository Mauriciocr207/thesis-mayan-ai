```python
from datetime import timedelta
```

```python
from kinai.models.video_annotation import Segment

segment = Segment(
  utt_id="asdasd",
  maya="ki'in",
  spanish="asdasd",
  start="00:00:00",
  end="00:00:00"
)

DIGITS_BASES = [
    3_600 * 10 * 1_000, 
    3_600 * 1_000, 
    60 * 10 * 1_000, 
    60 * 1_000, 
    10 * 1_000, 
    1 * 1_000, 
    100, 
    10, 
    1
]

def increment_digit(value: timedelta, pos: int):
    result = value + timedelta(milliseconds=DIGITS_BASES[pos])
  
    if result.days >= 1:
      return timedelta(days=1)
    
    return result 
    
def decrement_digit(value: timedelta, pos: int):
    result = value - timedelta(milliseconds=DIGITS_BASES[pos])
    
    print(result.days)
    if result.days < 0:
      return timedelta(microseconds=0)
  
    return result
    
for _ in range(10):
  segment.start_td = increment_digit(segment.start_td, 0)

def _timedelta_to_str(td: timedelta) -> str:
    total_ms = int(td.total_seconds() * 1000)

    hours = total_ms // (3600 * 1000)
    total_ms %= (3600 * 1000)

    minutes = total_ms // (60 * 1000)
    total_ms %= (60 * 1000)

    seconds = total_ms // 1000
    milliseconds = total_ms % 1000

    return f"{hours:02}:{minutes:02}:{seconds:02}.{milliseconds:03}"
  
segment.start, _timedelta_to_str(segment.start_td)
```
