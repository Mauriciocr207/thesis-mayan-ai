```python
import shutil
from pathlib import Path

from kinai.classes.paths import Paths

repeated = {}

def copy_audio(audio_path, new_name, destination_folder):
    destination_path = destination_folder / f"{new_name}.wav"
    shutil.copy(str(audio_path), str(destination_path))
    print(f"Copied {audio_path} to {destination_path}")

folder_1h_audio = Paths().segments_folder
folder_3h_audio = Path("./3h-audios")
```

```python
dest_audio_folder = Path("./audios")
dest_audio_folder.mkdir(parents=True, exist_ok=True)
```

```python
for i, row in df.iterrows():
    audio_path = folder_1h_audio / f"{row['filename']}.wav"
    if audio_path.exists():
        copy_audio(str(audio_path), row["utt_id"], dest_audio_folder)
        continue

    audio_path = folder_3h_audio / f"{row['filename']}.wav"
    if audio_path.exists():
        copy_audio(str(audio_path), row["utt_id"], dest_audio_folder)
        continue
    
    print(f"file {row['filename']} does not exist")
    print(i)
```

```python
from num2words import num2words
import re

def normalize_numbers(text):
    def replace_num(match):
        num = int(match.group())
        return num2words(num, lang='es')  # en español
    
    return re.sub(r'\d+', replace_num, text)

def normalize_text(text):
    # Primero números → palabras
    text = normalize_numbers(text)
    # Luego elimina los demás caracteres raros
    for char in {'$', ',', '-', '.', ':', ';', '¿'}:
        text = text.replace(char, '')
    return " ".join(text.split())

for idx, row in final_df.iterrows():
    maya = row["maya"]
    normalized = normalize_text(maya)

    if maya != normalized:
        print("-"*15)
        print(maya)
        print(normalized)
        print("-"*15)
```

```python
final_df_clean = final_df.copy()
final_df_clean["maya"] = final_df_clean["maya"].apply(normalize_text)
```
