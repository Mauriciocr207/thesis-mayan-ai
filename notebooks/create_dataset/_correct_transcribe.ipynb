```python
import os
import csv
import json
import re
from pathlib import Path
from typing import Tuple, Optional
import anthropic
import pandas as pd
import glob

from dotenv import load_dotenv

load_dotenv()

# Configuración
TRANSCRIPTS_CSV = "./data/3h-transcripts.csv"
TEXT_DIR = "./data/text/"
RESULTS_CSV = "./data/3h-corrected-transcripts.csv"

# Parámetros
INITIAL_WINDOW_WORDS = 100  # Palabras iniciales
MAX_WINDOW_WORDS = 100     # Máximo expandible
WINDOW_INCREMENT = 10      # Palabras para expandir

# Cliente Claude
client = anthropic.Anthropic(
    api_key=os.getenv("ANTHROPIC_API_KEY")
)

def write_csv(rows: list[dict]):
    file_exists = Path(RESULTS_CSV).exists()
    with open(RESULTS_CSV, "a" if file_exists else "w", newline="", encoding="utf-8") as f:
        fieldnames = ["file_name", "mms", "correction", "found"]
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        if not file_exists:
            writer.writeheader()
        writer.writerows(rows)

def load_transcriptions():
    if not Path(RESULTS_CSV).exists():
        rows = []
        with open(TRANSCRIPTS_CSV, newline="", encoding="utf-8") as f:
            for r in csv.DictReader(f):
                rows.append({
                    "file_name": r["audio_file"],
                    "mms": r["transcription"],
                    "correction": "",
                    "found": "no",
                })
        write_csv(rows)
        return rows

    with open(RESULTS_CSV, newline="", encoding="utf-8") as f:
        return list(csv.DictReader(f))

def load_reference_text(text_file: str) -> str:
    with open(text_file, 'r', encoding='utf-8') as f:
        return f.read().strip()

def save_transcriptions(rows: list[dict]):
    with open(RESULTS_CSV, "w", newline="", encoding="utf-8") as f:
        fieldnames = ["file_name", "mms", "correction", "found"]
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(rows)

def find_transcription_with_ai(
    window_text: str,
    mms_transcription:str,
) -> Tuple[Optional[str], int, bool]:
    prompt = f"""Tu tarea es recortar el texto de referencia.
Reglas obligatorias:
1. La respuesta debe comenzar exactamente en el primer carácter del texto de referencia.
2. La respuesta debe ser una copia exacta de los caracteres iniciales del texto de referencia.
3. Solo debes decidir en qué punto detener la copia.
4. No puedes modificar, corregir, reordenar ni agregar texto.
5. No puedes usar texto del segmento si no aparece exactamente en el texto de referencia.
6. Si no puedes identificar con claridad el punto final, responde exactamente: NO_ENCONTRADO

Segmento transcrito:
{mms_transcription}

Texto de referencia:
{window_text}

Devuelve únicamente el recorte exacto del inicio del texto de referencia.
No expliques nada.
No uses comillas.
"""

    try:
        message = client.messages.create(
            model="claude-opus-4-6",
            max_tokens=256,
            messages=[
                {"role": "user", "content": prompt}
            ]
        )
        
        response = message.content[0].text.strip()
        
        return response

    except Exception as e:
        print(f"    ✗ Error: {e}")
```

```python
def process_transcripts():
    """Procesa todas las transcripciones del CSV."""
    transcriptions = load_transcriptions()
    print(f"✓ Cargadas {len(transcriptions)} transcripciones")
    founded_segments = sum(map(lambda x: x["found"] == "yes", transcriptions))
    print(f"✓ {founded_segments} segmentos ya procesados")

    textos = {}
    for text_file in sorted(glob.glob("./text/*.txt")):        
        # Cargar texto de referencia
        file_name = Path(text_file).stem
        text_file = load_reference_text(text_file)
        textos[file_name] = text_file.strip()

    for idx, transcription in list(enumerate(transcriptions)):
        file_name = transcription["file_name"]

        copy = textos.get(file_name, "").strip()

        # 🔹 Si ya fue procesado, recorta el texto automáticamente
        if transcription["found"] == "yes" or not "15_Mario_Chan" in file_name:
            correction = transcription.get("correction", "").strip()
            copy = copy.strip()

            print(f"[CORR]{correction}")
            print(f"[COPY]{copy}")

            if correction and copy.startswith(correction):
                textos[file_name] = copy[len(correction):].strip()
            else:
                print(f"[WARN] No coincide corrección previa en {file_name}_{idx+1}")

            print(f"[SKIP] {file_name}_{idx+1}")
            continue

        window = copy[:220]
        founded = find_transcription_with_ai(window, transcription["mms"])

        if founded and founded != "NO_ENCONTRADO":
            founded = founded.strip()
            copy = copy[len(founded):]
            transcription["correction"] = founded
            transcription["found"] = "yes"
        else:
            transcription["correction"] = ""
            transcription["found"] = "no"

        save_transcriptions(transcriptions)

        print(f"""
[SEG]       {file_name}_{(idx)+1}
[REM LEN]   {len(copy)}
[REM TXT]   {window}
[MMS]       {transcription["mms"]}
[RES]       {founded}
""")

        textos[file_name] = copy

if __name__ == "__main__":
    process_transcripts()
```
