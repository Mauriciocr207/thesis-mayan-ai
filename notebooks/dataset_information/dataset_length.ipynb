```python
import csv
import subprocess
import json
from pathlib import Path

def get_duration(wav_path):
    cmd = [
        "ffprobe", "-v", "error",
        "-show_entries", "format=duration",
        "-of", "json",
        wav_path
    ]
    result = subprocess.run(cmd, capture_output=True, text=True)
    data = json.loads(result.stdout)
    return float(data["format"]["duration"])

csv_path = str(Path(get_root()) / "data/01_raw/spoken_dictionary_registry.csv")
total_duration = 0.0
count = 0

with open(csv_path, newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    print("Cargando información...")
    for row in reader:
        wav = row["audio_path"]

        if not Path(wav).exists():
            print(f"⚠️ No existe: {wav}")
            continue

        dur = get_duration(wav)
        total_duration += dur
        count += 1
print("\nResumen")
print("Audios:", count)
print(f"Total segundos: {total_duration:.2f}")
print(f"Total minutos: {total_duration/60:.2f}")
print(f"Total horas: {total_duration/3600:.4f}")
```
