```python
from pathlib import Path

def gen_text_files(dir: str):
    input_dir = Path(dir)
    for file in input_dir.rglob("*"):
        if file.suffix.lower() in [".wav", ".mp3", ".flac", ".m4a"]:
            filename = file.name.split(".wav")[0]
            with open(NARRACIONES_TRANSCRIPTS / f"{filename}.txt", "w+") as f:
                f.write("")

gen_text_files(NARRACIONES_MP3)
```
