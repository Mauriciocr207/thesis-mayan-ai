```python
import requests

for url in urls:
    file = url.split("narraciones_mayas_campeche/")[1]
    filename = file.split(".mp3")[0]
    with requests.get(url, stream=True) as r:
        with open(NARRACIONES_MP3 / f"{filename}.mp3", "wb") as f:
            for chunk in r.iter_content(chunk_size=8192):
                f.write(chunk)
```

```python
import subprocess
from pathlib import Path

def convert_mp3_to_wav(input_dir, output_dir):
    input_dir = Path(input_dir)
    output_dir = Path(output_dir)

    for mp3_file in input_dir.rglob("*.mp3"):
        # Mantener estructura de carpetas
        relative_path = mp3_file.relative_to(input_dir).with_suffix(".wav")
        wav_file = output_dir / relative_path

        wav_file.parent.mkdir(parents=True, exist_ok=True)

        cmd = [
            "ffmpeg",
            "-y",
            "-i", str(mp3_file),
            "-ar", "16000",
            "-ac", "1",
            "-c:a", "pcm_s16le",
            str(wav_file)
        ]

        try:
            subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
            print(f"✔ {mp3_file} → {wav_file}")
        except subprocess.CalledProcessError:
            print(f"✘ Error en {mp3_file}")

# === USO ===
convert_mp3_to_wav(
    input_dir=NARRACIONES_MP3,
    output_dir=WORK / "narraciones_wav"
)
```

```python
import subprocess
from pathlib import Path
from datetime import timedelta
import re

def get_duration(file_path):
    cmd = [
        "ffprobe",
        "-v", "error",
        "-show_entries", "format=duration",
        "-of", "default=noprint_wrappers=1:nokey=1",
        str(file_path)
    ]
    result = subprocess.run(cmd, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
    
    try:
        return float(result.stdout.strip())
    except:
        return 0.0
    
def extract_number(path):
    match = re.search(r"/(\d+)_", str(path))
    if match:
        return int(match.group(1))
    return float("inf")

def format_duration(seconds):
    td = timedelta(seconds=seconds)
    
    total_seconds = int(td.total_seconds())
    ms = int((td.total_seconds() - total_seconds) * 1000)
    
    hours = total_seconds // 3600
    minutes = (total_seconds % 3600) // 60
    secs = total_seconds % 60
    
    return f"{hours:02d}:{minutes:02d}:{secs:02d}.{ms:03d}"

def total_audio_duration(directory):
    directory = Path(directory)

    total_seconds = 0.0
    count = 0

    results = []

    for file in directory.rglob("*"):
        if file.suffix.lower() in [".wav", ".mp3", ".flac", ".m4a"]:
            duration = get_duration(file)
            total_seconds += duration
            count += 1

            results.append((file, duration))

    # 🔥 ORDENAR POR ID (01, 02, 03...)
    results.sort(key=lambda x: extract_number(x[0]))

    for file, duration in results:
        print(f"{file} -- {format_duration(duration)}")

    minutes = total_seconds / 60
    hours = total_seconds / 3600

    print("\n===== RESUMEN =====")
    print(f"Archivos procesados: {count}")
    print(f"Total: {minutes:.2f} minutos")
    print(f"Total: {hours:.2f} horas")

    return results

# === USO ===
results = total_audio_duration(WORK / "narraciones_wav")
```
