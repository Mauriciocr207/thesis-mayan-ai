```python
import subprocess
import numpy as np
import soundfile as sf
from pathlib import Path
import subprocess
from pathlib import Path

# Fuente: los wav que convierte 00a_narraciones_download desde los mp3 del INALI.
# Cortar desde el wav (y no desde el mp3) es lo que reproduce el corpus byte a byte:
# ffmpeg con -ss antes de -i busca por frames en mp3 y el decodificado introduce
# unas pocas muestras de diferencia por redondeo.
narraciones_folder = WORK / "narraciones_wav"
dest_folder = WORK / "3h-audios"

dest_folder.mkdir(parents=True, exist_ok=True)

def segment_audio(audio_name, utt_id, start, end):
    audio_path = dest_folder / f"{utt_id}.wav"

    if audio_path.exists():
        print(f"[skip] {utt_id}")
        return

    start = f"{float(start):.3f}"
    end = f"{float(end):.3f}"

    cmd = [
        "ffmpeg",
        "-y",
        "-ss", start,
        "-to", end,
        "-i", str(narraciones_folder / f"{audio_name}.wav"),
        "-ac", "1",
        "-ar", "16000",
        str(audio_path)
    ]

    print(f"[segmenting] {utt_id} ({start} → {end})")
    result = subprocess.run(cmd, capture_output=True, text=True)

    if result.returncode != 0:
        print("STDERR:")
        print(result.stderr)
        raise RuntimeError(f"ffmpeg falló con código {result.returncode}")

    normalize_peak(audio_path)
    print(f"[done] {utt_id}")

def normalize_peak(path: Path, target: float = 0.99):
    data, sr = sf.read(path)
    peak = np.max(np.abs(data))
    if peak > 0:
        data = data * (target / peak)
    sf.write(path, data, sr, subtype="PCM_16")
```

```python
import pandas as pd
df = pd.read_csv(ANNOTATIONS / "3h-transcripts.csv")
```

```python
for _, row in df.iterrows():
    audio_file = row["audio_file"]
    segment_id = row["segment_id"]
    time_start = str(row["time_start"])
    time_end = str(row["time_end"])
    duration = row["duration"]

    segment_audio(
        audio_name=audio_file,
        utt_id=f"{audio_file}_{segment_id}",
        start=time_start,
        end=time_end,
    )
```
