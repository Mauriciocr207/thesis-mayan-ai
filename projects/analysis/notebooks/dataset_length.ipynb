```python
import json
import subprocess


def get_duration(wav_path):
    cmd = ["ffprobe", "-v", "error", "-show_entries", "format=duration",
           "-of", "json", str(wav_path)]
    result = subprocess.run(cmd, capture_output=True, text=True)
    return float(json.loads(result.stdout)["format"]["duration"])


total_duration = 0.0
count = 0

print("Cargando información...")
for path in df["path"]:
    wav = AUDIO / Path(path).name
    if not wav.exists():
        print(f"⚠️ No existe: {wav}")
        continue
    total_duration += get_duration(wav)
    count += 1

print("\nResumen")
print("Audios:", count)
print(f"Total segundos: {total_duration:.2f}")
print(f"Total minutos: {total_duration/60:.2f}")
print(f"Total horas: {total_duration/3600:.4f}")
```
